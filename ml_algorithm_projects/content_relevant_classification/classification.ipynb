```r
#1. Load data 

ind_train = read.csv("forum_ind_train.csv", stringsAsFactors = FALSE)
dep_train = read.csv("forum_dep_train.csv", stringsAsFactors = FALSE)
cat("Data loaded\n")
```

```r
head(ind_train,5)
head(dep_train,5)
```

```r
#2. Debug the dataset - for joinning later 

#check dimension 
cat("ind_train dim:" , dim(ind_train),"\n")
cat("dep_train dim:" , dim(dep_train),"\n")

# check Unique_ID col
cat("Unique_ID in independant train :" , length(unique(ind_train$Unique_ID)), "out of", nrow(ind_train),"\n")
cat("Unique_ID in dependant train :" , length(unique(dep_train$Unique_ID)), "out of", nrow(dep_train),"\n")
```

```r
# check for missing Ids
missing_ids_ind = setdiff(ind_train$Unique_ID, dep_train$Unique_ID)
cat("Ids in independant missing from dependant:", length(missing_ids_ind), "\n")
print(head(missing_ids_ind,5))

missing_ids_dep = setdiff(dep_train$Unique_ID, ind_train$Unique_ID)
cat("Ids in dependant missing from independant:", length(missing_ids_dep), "\n")
print(head(missing_ids_dep,5))

#check for duplicated Ids 
cat("duplicated Ids in dependant:", sum(duplicated(dep_train$Unique_ID)), "\n")
cat("duplicated Ids in independant:", sum(duplicated(ind_train$Unique_ID)), "\n")
```

```r
# check for label column in dependant 
cat("unique labels in dependant data :", unique(dep_train$label), "\n")
cat("NA values in labels:", sum(is.na(dep_train$label)),"\n")
```

```r
#3. Merging data (join data)

#remove duplicated and NA 

ind_train = ind_train[!duplicated(ind_train$Unique_ID),]
dep_train = dep_train[!duplicated(dep_train$Unique_ID),]

# remove NA labels from dependant data
dep_train = dep_train[!is.na(dep_train$label),]
#check dimension 
cat("ind_train dim:" , dim(ind_train),"\n")
cat("dep_train dim:" , dim(dep_train),"\n")
```

```r
# merge data 

train_data = merge(ind_train, dep_train, by = "Unique_ID")
cat("train data dim:" , dim(train_data),"\n")
all_cols = colnames(train_data)
feature_cols = setdiff(all_cols, 
                       c("Unique_ID", "unit_faculty", "demographic_sex", "label")
                       )

cat("total features:", length(feature_cols), "\n")
```

```r
#4. load validation data 

ind_val = read.csv("forum_ind_validation.csv", stringsAsFactors = FALSE) #treat strings as factor in dataset rather than just string
dep_val = read.csv("forum_dep_validation.csv", stringsAsFactors = FALSE)
```

```r
head(ind_val,5)
head(dep_val,5)
```

```r
#5. Function to debug/check data 

debug_dataset = function(ind_val, dep_val) {

    #check if col name exist 
    cat("Cols existence check\n")
    if(!"Unique_ID" %in% names(ind_val)) { stop("error:cols not exist in indepedant") } else {cat("valid")}

    if(!"Unique_ID" %in% names(dep_val)) { stop("error:cols not exist dependant") } else {cat("valid")}
    
    if(!"label" %in% names(dep_val)) { stop("error:cols not exist dependant") } else {cat("valid")}

    cat("Dataset check\n")

    cat("ind data dim:", dim(ind_val),"\n")
    cat("dep data dim:", dim(dep_val),"\n")

    # check Unique_ID col
    cat("Unique_ID in independant val :" , length(unique(ind_val$Unique_ID)), "out of", nrow(ind_val),"\n")
    cat("Unique_ID in dependant val :" , length(unique(dep_val$Unique_ID)), "out of", nrow(dep_val),"\n")

    # check for missing Ids
    missing_ids_ind = setdiff(ind_val$Unique_ID, dep_val$Unique_ID)
    cat("Ids in independant missing from dependant:", length(missing_ids_ind), "\n")

    missing_ids_dep = setdiff(dep_val$Unique_ID, ind_val$Unique_ID)
    cat("Ids in dependant missing from independant:", length(missing_ids_dep), "\n")

    #check for duplicated Ids 
    cat("duplicated Ids in dependant:", sum(duplicated(dep_val$Unique_ID)), "\n")
    cat("duplicated Ids in independant:", sum(duplicated(ind_val$Unique_ID)), "\n")

    #merge verification

    merge_result = merge(ind_val,dep_val, by = "Unique_ID", all = FALSE)
    cat("merge result dim :", dim(merge_result), "\n")

    invisible(list(
        ind_val_dim = dim(ind_val),
        dep_val_dim = dim(dep_val),
        merge_dim = dim(merge_result)
        ))    
    
    }
```

```r
debug_dataset(ind_val, dep_val)
```
